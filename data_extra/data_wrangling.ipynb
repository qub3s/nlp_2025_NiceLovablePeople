```python
df = pd.read_csv('IMDB Dataset.csv')
df.shape
```

```python
df.head(2)
```

```python
df = df.reset_index()
```

```python
df.columns = ['id', 'sentence', 'sentiment']
df['id'] = 'xxx-' + df['id'].astype(str) + '-xxx'  
df.tail(2)
```

```python
df = df.replace({'sentiment': {'positive': 4, 'negative': 0}})
df.sample(5)
```

```python
# df = df.sample(frac=0.2, random_state=11711).reset_index(drop=True)  # shuffle the dataframe
# df.shape
```

```python
df = df[df.sentence.apply(lambda x: len(x.split())) <= 56].reset_index(drop=True)  # filter sentences with more than 56
df.shape
```

```python
# find odd sentences which may break bert
df.sentence.apply(lambda x: len(x.split())).describe()
```

```python
# append df to original_df
df = pd.concat([original_df, df], axis=0).reset_index(drop=True) 
df
```

```python
df.to_csv('../data/sst-sentiment-train.csv', index=False)
```
